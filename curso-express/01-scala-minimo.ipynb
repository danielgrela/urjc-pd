## Funciones

- `def nombre[TiposGenéricos](param: Tipo): TipoResultado = cuerpo`
- **No hay `return`**: el resultado es la última expresión.
- El cuerpo va indentado (como en Python).

```scala
// def      → estoy definiendo una función
// doble    → su nombre
// (n: Int) → recibe un parámetro n de tipo Int (en Scala el tipo va DETRÁS del nombre)
// : Int    → devuelve un Int
// =        → a continuación viene el cuerpo
def doble(n: Int): Int =
    n * 2          // no hay "return": la última expresión es lo que se devuelve

// [A] declara un "tipo cualquiera" (como <T> en Java).
// Así la misma función vale para Int, String, listas... Scala deduce A al llamarla.
def identidad[A](a: A): A = a

doble(21)          // el notebook muestra el valor de la última línea de la celda
```

## Lambdas (funciones anónimas)

`A => B` es el **tipo** "función de A en B". Una lambda se escribe `x => cuerpo`.
`_ + 1` es abreviatura de `x => x + 1` (cada `_` es un parámetro distinto).
`A => B => C` es una función que devuelve otra función: se llama `f(a)(b)`.

```scala
// Tipo Int => Int: "función que recibe un Int y devuelve un Int".
// A la izquierda de la flecha va el parámetro (x) y a la derecha el resultado (x + 1).
val inc: Int => Int = x => x + 1

// _ + _ es la forma corta de (a, b) => a + b: el primer _ es el primer parámetro
// y el segundo _ el segundo. ¡No significa usar el mismo valor dos veces!
val suma: (Int, Int) => Int = _ + _

// Int => Int => Int se lee Int => (Int => Int): recibe a y DEVUELVE otra función
// que espera b. Por eso se llama en dos pasos: sumaCurry(2)(3).
// Esta forma "currificada" es la que aparece en Curry-Howard: p => q => ...
val sumaCurry: Int => Int => Int = a => b => a + b

(inc(1), suma(2, 3), sumaCurry(2)(3))   // una tupla para ver los tres resultados a la vez
```

## Tuplas, `Option` y `Either`

| Tipo | Valores | Significado |
|---|---|---|
| `(A, B)` | `(a, b)`; se accede con `._1`, `._2` | una A **y** una B |
| `Option[A]` | `Some(a)` o `None` | una A o nada |
| `Either[A, B]` | `Left(a)` o `Right(b)` | una A **o** una B |
| `Unit` | `()` | un único valor |
| `Nothing` | ninguno | imposible (el "falso" lógico) |

## Pattern matching: `match`

Mira la forma de un valor y saca sus piezas. Los casos se prueban en orden; `_` es "cualquier cosa"; `if` añade una condición.

```scala
def describe(e: Either[Int, String]): String =
    e match                                   // miro qué forma tiene e
        // Left(n) encaja si e es un Left, y además le da el nombre n a lo que hay dentro.
        // "if n < 0" es una guarda: este caso solo se aplica si además se cumple la condición.
        case Left(n) if n < 0 => "número negativo"
        // Este caso va DESPUÉS porque es más general: los casos se prueban de arriba abajo
        // y gana el primero que encaje. Si lo pusiera antes, el de negativos nunca se usaría.
        case Left(n)          => s"número $n"     // s"...$n" mete el valor de n en el texto
        case Right(s)         => s"texto $s"
        // No hace falta un caso por defecto: Either solo puede ser Left o Right.

describe(Left(3))
```

Una función que solo hace `match` sobre su argumento se puede escribir directamente con los `case`:

```scala
// En lugar de  o => o match { case ... }  se escriben directamente los case:
// Scala sabe por el tipo declarado (Option[Int] => Boolean) qué reciben.
// Esta forma se usa muchísimo en el examen (Curry-Howard, isomorfismos, folds).
val esCero: Option[Int] => Boolean =
    case Some(0) => true      // encaja solo con un Some que contenga exactamente 0
    case _       => false     // _ = "cualquier otra cosa" (None, Some(5)...)

esCero(Some(0))
```

## Listas

Una lista es `Nil` (vacía) o `cabeza :: resto`. Así se recorre **siempre**:

```scala
// Una lista es O BIEN Nil (vacía) O BIEN head :: tail (un primer elemento y el resto).
// Como el resto es otra lista, la función se llama a sí misma sobre él (recursión).
def suma(l: List[Int]): Int =
    l match
        case Nil => 0                          // caso base: la suma de nada es 0
        case head :: tail =>                   // separo el primero (head) del resto (tail)
            head + suma(tail)                  // primero + (lo que sume el resto)
            // No hace falta "seguir" la recursión mentalmente: confía en que suma(tail)
            // ya da la suma correcta del resto, y piensa solo en este paso.

suma(List(1, 2, 3))    // = 1 + suma(List(2,3)) = 1 + (2 + suma(List(3))) = 1 + 2 + 3 + 0
```

Útiles: `l.reverse`, `l1 ++ l2`, `l :+ x` (añadir al final), `l.contains(x)`, `l.length`, `l.take(n)`, `l.drop(n)`, `l.splitAt(n)`, `l.zip(otra)`, `l.zipWithIndex`, `l.sum`.

⚠️ Al escribir una lista vacía como valor inicial, **dile el tipo**: `Nil: List[Int]` o `List[Int]()`. Si no, Scala cree que es una lista de "nada" y luego no te deja meterle elementos.
