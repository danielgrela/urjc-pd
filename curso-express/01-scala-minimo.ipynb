## Funciones

- `def nombre[TiposGenéricos](param: Tipo): TipoResultado = cuerpo`
- **No hay `return`**: el resultado es la última expresión.
- El cuerpo va indentado (como en Python).

```scala
def doble(n: Int): Int =
    n * 2

def identidad[A](a: A): A = a      // [A] = "para cualquier tipo A"

doble(21)
```

## Lambdas (funciones anónimas)

`A => B` es el **tipo** "función de A en B". Una lambda se escribe `x => cuerpo`.
`_ + 1` es abreviatura de `x => x + 1` (cada `_` es un parámetro distinto).
`A => B => C` es una función que devuelve otra función: se llama `f(a)(b)`.

```scala
val inc: Int => Int = x => x + 1
val suma: (Int, Int) => Int = _ + _
val sumaCurry: Int => Int => Int = a => b => a + b

(inc(1), suma(2, 3), sumaCurry(2)(3))
```

## Tuplas, `Option` y `Either`

| Tipo | Valores | Significado |
|---|---|---|
| `(A, B)` | `(a, b)`; se accede con `._1`, `._2` | una A **y** una B |
| `Option[A]` | `Some(a)` o `None` | una A o nada |
| `Either[A, B]` | `Left(a)` o `Right(b)` | una A **o** una B |
| `Unit` | `()` | un único valor |
| `Nothing` | ninguno | imposible (el "falso" lógico) |

## Pattern matching: `match`

Mira la forma de un valor y saca sus piezas. Los casos se prueban en orden; `_` es "cualquier cosa"; `if` añade una condición.

```scala
def describe(e: Either[Int, String]): String =
    e match
        case Left(n) if n < 0 => "número negativo"
        case Left(n)          => s"número $n"
        case Right(s)         => s"texto $s"

describe(Left(3))
```

Una función que solo hace `match` sobre su argumento se puede escribir directamente con los `case`:

```scala
val esCero: Option[Int] => Boolean =
    case Some(0) => true
    case _       => false

esCero(Some(0))
```

## Listas

Una lista es `Nil` (vacía) o `cabeza :: resto`. Así se recorre **siempre**:

```scala
def suma(l: List[Int]): Int =
    l match
        case Nil => 0
        case head :: tail => head + suma(tail)

suma(List(1, 2, 3))
```

Útiles: `l.reverse`, `l1 ++ l2`, `l :+ x` (añadir al final), `l.contains(x)`, `l.length`, `l.take(n)`, `l.drop(n)`, `l.splitAt(n)`, `l.zip(otra)`, `l.zipWithIndex`, `l.sum`.

⚠️ Al escribir una lista vacía como valor inicial, **dile el tipo**: `Nil: List[Int]` o `List[Int]()`.
