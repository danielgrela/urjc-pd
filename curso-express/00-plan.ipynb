# Curso express · Aprobar Programación Funcional en ~2 horas

Este curso está pensado para **una sola sesión de 1h45m - 2h**. No intenta enseñarlo todo:
se centra en lo que **sale en todos los exámenes** y en recetas que se pueden aplicar casi
sin pensar. Si tienes más tiempo, el temario completo está en la carpeta `claudetemario`.

## El examen en una tabla

| Ejercicio | Puntos | Qué es | ¿Cuánto sale? | Dificultad con receta |
|---|---|---|---|---|
| 1 | 2 | Curry-Howard: demostrar una fórmula programando | **siempre** | baja |
| 2 | 1-2 | Isomorfismo entre dos tipos | casi siempre | baja |
| 3 | 3-4 | Una función de listas de 3-5 formas (recursión, `foldRight`, recursión final, `foldLeft`, `map`/`filter`/`flatMap`) | casi siempre | media |
| 4 | 3 | Abstraer un patrón común en una función de orden superior (o árboles) | casi siempre | media |

**Para aprobar (5 puntos):** Ejercicios 1 y 2 enteros (≈ 4 puntos, son mecánicos) + las
partes de recursión y `foldRight` del Ejercicio 3 (≈ 1,5-2 puntos). Todo lo demás es nota.

**Cómo usar cada notebook:** lee la receta, ejecuta los ejemplos (primero la celda del
preámbulo) y haz los mini-ejercicios **antes** de desplegar la solución.

## Estrategia el día del examen

1. **Lee el preámbulo** del notebook del examen: te dice qué funciones puedes usar.
2. **Empieza por el Ejercicio 1 y luego el 2.** Son los más rápidos con receta.
3. En el 3 y el 4, haz **primero el apartado recursivo**: los demás (`foldRight`, `foldLeft`…) se sacan traduciéndolo.
4. **Ejecuta los tests** de cada apartado. Si fallan, lee qué caso falla y prueba ese ejemplo a mano.
5. **No dejes nada en blanco**: una solución parcial que compila puntúa más que nada.
6. Respeta lo que pide cada apartado ("con recursión final", "solo con `filter`"…), aunque otra forma te parezca más fácil.
