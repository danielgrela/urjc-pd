# claudetemario · Cursillos para el examen de Programación Funcional (Scala)

Material generado a partir de los **22 exámenes resueltos** de `exams/` (cursos 19-20 a
25-26) y de los apuntes de `PF-1`, `PF-2` y `PF-3`. Está pensado para alguien que **nunca
ha tocado Scala**: cada cosa nueva de la sintaxis se explica la primera vez que aparece
(sobre todo en el archivo 01).

---

## Cómo es el examen

- **Duración:** 1h20m - 1h30m. Se hace en un notebook de Jupyter con Scala 3.
- **Estructura casi fija** (cursos 21-22 a 25-26):

| Ejercicio | Puntos | Qué te piden | Cursillo |
|---|---|---|---|
| 1 | 2 | Demostrar una fórmula lógica escribiendo un programa (a: intuicionista, b: clásica con tercio excluso / doble negación) | **02** |
| 2 | 1-2 | Demostrar que dos tipos son isomorfos (a: con aritmética, b: con `from`/`to`) | **03** |
| 3 | 3-4 | Una función sobre listas, implementada de varias formas: recursión, `foldRight`, recursión final, `foldLeft`, `map`/`filter`/`flatMap` | **04, 05, 06** |
| 4 | 3-3,5 | Abstraer un patrón común a varias funciones en una función de orden superior (o un ejercicio de árboles) | **08** (y **07**) |

- Al principio del notebook hay un **preámbulo** con definiciones que puedes usar (`Tree`,
  `Isomorphic`, `Not`, `ExcludedMiddle`, `DoubleNegation`, signaturas de `List`…). Léelo
  antes de empezar.
- Cada apartado trae **tests**: ejecútalos (`run(TestX(miFuncion))`). Si pasan, vas bien.

## Plan de estudio recomendado

1. **01** completo (una vez). Vuelve a él como diccionario.
2. **02** y **03**: son mecánicos y valen 3-4 puntos juntos. Haz todos los ejercicios.
3. **04** → **05** → **06**: en este orden, porque cada uno se apoya en el anterior.
4. **08** y luego **07**.
5. **Simulacro:** abre un examen con `-Enunciado` en el nombre (por ejemplo
   `exams/Examen25-26-Ordinaria-Móstoles-Enunciado.ipynb`), hazlo con reloj, y corrige con
   el mismo archivo sin `-Enunciado`. Empieza por los de 25-26 y 24-25.

### Erratas detectadas en las soluciones de `exams/`

- **25-26 Vicálvaro Bis**: las soluciones de los ejercicios 1b, 2, 3 y 4 son las del examen
  de Vicálvaro (otra fórmula / otros tipos / otro enunciado). Las correctas están en los
  cursillos 02 (ejemplo 13), 03 (ejemplo 3), 05 (`split` con recursión final y `foldLeft`)
  y 08 (`mapLeft`).
- **23-24 Extraordinaria**: la función `tails` de ese examen (en el anexo) **no** incluye la
  lista completa, al contrario que la de 23-24 Móstoles. Ver cursillo 06.
