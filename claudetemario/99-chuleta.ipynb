## Curry-Howard (Ejercicio 1)

| Lógica | Tipo | Construir | Usar |
|---|---|---|---|
| p → q | `P => Q` | `p => ...` | `f(p)` |
| p ∧ q | `(P, Q)` | `(p, q)` | `._1`, `._2`, `case (p, q)` |
| p ∨ q | `Either[P, Q]` | `Left(p)` / `Right(q)` | `match` |
| ¬p | `P => Nothing` | `p => (Nothing)` | `np(p): Nothing` |
| ⊥ | `Nothing` | aplicar una negación | vale como cualquier tipo |

- Precedencia: ¬ > ∧ > ∨ > → ; la → asocia a la **derecha**. `p ∧ q → r` = `((P, Q)) => R`.
- Premisas `{…} ⊢ c` → parámetros de la función.
- LEM: `lem[X] match { case Left(x) => …; case Right(nx) => … }`
- DN: meta `X` → `dn[X](nx => … nx(algo: X) …)`

## Listas (Ejercicio 3)

```scala
// Recursión (divide y vencerás)          // foldRight
l match                                   l.foldRight(BASE)((head, tailSol) => COMBINAR)
    case Nil => BASE
    case head :: tail =>
        val tailSol = f(tail)
        COMBINAR(head, tailSol)

// Recursión final (iterativo)             // foldLeft
def step(out: B, aux: List[A]): B =       l.foldLeft(INICIAL)((out, head) => ACTUALIZAR)
    aux match                             // + post-proceso (.reverse, ._1, match)
        case Nil => out
        case head :: tail => step(ACTUALIZAR(out, head), tail)
step(INICIAL, l)  // + post-proceso
```

- `foldRight`: `(elemento, solResto)`. `foldLeft`: `(acumulado, elemento)`.
- Con `foldLeft` + `::` la lista sale **al revés** → `.reverse`.
- Si te falta información → el acumulado es una **tupla** (índice, anterior `Option`, flag…) y al final `._1`.
- Pattern matching dentro del fold: `l.foldRight(z): case (h, sol) => ...` (Scala 2: `{ case ... }`).
- `(head, f(tail)) match` para enumerar combinaciones.
